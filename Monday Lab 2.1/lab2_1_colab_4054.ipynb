# **Task 1**
**Prompt**:
You are an AI coding assistant.
Write a Python function in Google Colab that:
Accepts a list of numbers as input
Calculates the mean, minimum, and maximum values
Prints the results clearly

Then run the function on the list [12, 7, 9, 21, 5] and show the output.

```python
def analyze_numbers(numbers):
    """
    Calculates and prints the mean, minimum, and maximum values of a list of numbers.

    Args:
        numbers (list): A list of numerical values.
    """
    if not numbers:
        print("The input list is empty.")
        return

    mean_value = sum(numbers) / len(numbers)
    min_value = min(numbers)
    max_value = max(numbers)

    print(f"Input List: {numbers}")
    print(f"Mean: {mean_value:.2f}")
    print(f"Minimum: {min_value}")
    print(f"Maximum: {max_value}")

# Run the function on the list [12, 7, 9, 21, 5]
my_list = [12, 7, 9, 21, 5]
analyze_numbers(my_list)
```

# **Task 2**
**Prompt :**
You are an AI coding assistant.
Write a Python function in Google Colab that:
Checks if a given number is an Armstrong number
Prints whether the number is Armstrong or not
Then run the function on the numbers 153 and 123 and show the output.

```python
def is_armstrong_number(number):
    """
    Checks if a given number is an Armstrong number and prints the result.

    An Armstrong number (also known as a narcissistic number) is a number
    that is the sum of its own digits each raised to the power of the
    number of digits.

    Args:
        number (int): The number to check.
    """
    num_str = str(number)
    n = len(num_str)
    sum_of_powers = 0
    for digit in num_str:
        sum_of_powers += int(digit) ** n

    if sum_of_powers == number:
        print(f"{number} IS an Armstrong number.")
    else:
        print(f"{number} IS not an Armstrong number.")

# Run the function on the numbers 153 and 123
is_armstrong_number(153)
is_armstrong_number(123)
```
